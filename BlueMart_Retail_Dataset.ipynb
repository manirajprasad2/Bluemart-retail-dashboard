* **Sales Trends:** How are sales changing over time?

```python
monthly_sales=df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales.index=monthly_sales.index.astype(str)
```

```python
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker='o',color='blue')
plt.title('Monthly sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=100)
plt.tight_layout()
plt.show
```

* **Category Performance:** Which product categories generate the highest sales and profits?

# profit=total value-quantity * cost price

```python
df['profit']=df['total_value']-(df['quantity']* df['cost_price'])
category_perf=df.groupby('category').agg(total_sales=('total_value','sum'),
                           total_profit=('profit','sum')).sort_values('total_sales',ascending=False)

category_perf
```

```python
# Calculate profit for each transaction
df['profit'] = df['total_value'] - (df['quantity'] * df['cost_price'])

# Calculate sales and profit by category
category_perf = df.groupby('category').agg(
    total_sales=('total_value', 'sum'),
    total_profit=('profit', 'sum')
).sort_values('total_sales', ascending=False)

# Display the result
print(category_perf)
```

```python
# Visualization--bar chart
x=np.arange(len(category_perf))
width=0.35
plt.figure(figsize=(12,5))
plt.bar(x-width/2,category_perf['total_sales'],width,label='Total Sales',color='green')
plt.bar(x+width/2,category_perf['total_profit'],width,label='Total Profit',color='red')
plt.xticks(x,category_perf.index,rotation=90)
plt.ylabel('Amount')
plt.xlabel('Categoryy')
plt.title('Category-wise Sales vs Profit')
plt.legend()
plt.tight_layout()
plt.show()
```

* **Product Profitability:** Which products contribute the most to overall profit?

```python
product_perf=df.groupby('subcategory').agg(total_sales=('total_value','sum'),
                                          total_profit=('profit','sum'))

product_perf=product_perf.sort_values('total_profit',ascending=False)

top10_profitable=product_perf.head(10)
top10_profitable
```

```python
#Visualization--bar chart
plt.figure(figsize=(11,5))
plt.bar(top10_profitable.index,top10_profitable['total_profit'],width,label='Total Profit',color='blue')
plt.xlabel('Product')
plt.ylabel('Total Profit')
plt.title('Top 10 Most Profitable Products')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show
```

* **Store & City Performance:** Which stores and cities deliver the best sales performance?

```python
store_perf=df.groupby('store_name')['total_value'].sum().sort_values(ascending=False)
store_perf

# Visualization-bar chart
plt.figure(figsize=(13,5))
plt.bar(store_perf.index,store_perf.values,color='purple')
plt.xlabel('store')
plt.ylabel('Total sales')
plt.title('store wise salse performance')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


```

```python
city_perf=df.groupby('city')['total_value'].sum().sort_values(ascending=False)
plt.figure(figsize=(13,5))
plt.bar(city_perf.index,city_perf.values,color='red')
plt.xlabel('city')
plt.ylabel('Total sales')
plt.title('city wise sales performance')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

* **Sales Channel Analysis:** Which sales channels contribute the most revenue?

```python
channel_perf=df.groupby('channel')['total_value'].sum().sort_values(ascending=False)
channel_perf
#Visualization --bar chart
plt.figure(figsize=(8,5))
plt.bar(channel_perf.index,channel_perf.values,color='green')
plt.xlabel('channel')
plt.ylabel('Total Sales')
plt.title('Channel wise Revenue Contribution')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Which high selling products are at risk of stockout(low inventory)?
```

```python
Product_sales=df.groupby("sku_id")['quantity'].sum().reset_index()

Product_sales
```

```python
risk_df=Product_sales.merge(idf,on='sku_id',how='left')
risk_df
```

```python
high_sales_low_stock=risk_df[(risk_df['quantity']>risk_df['quantity'].median()) & (risk_df['stock_on_hand']<risk_df['stock_on_hand'].median())]
```

3 Which product have high sales but low profit?

```python

product_summary = df.groupby('sku_name').agg(
    total_sales=('total_value', 'sum'),
    total_profit=('profit', 'sum')
)

product_summary

```

```python

high_sales_low_profit = product_summary.loc[
    (product_summary['total_sales'] > product_summary['total_sales'].median()) &
    (product_summary['total_profit'] < product_summary['total_profit'].median())
]

high_sales_low_profit
```

Sales-->Relatively High & Profit--> Relatively low
